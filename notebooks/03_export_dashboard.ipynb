```python
import pandas as pd

df = pd.read_csv("../data/bs140513_032310.csv")

# Export 1 - Bilan des regles de gestion
bilan_regles = pd.DataFrame({
    "Regle": ["RG01", "RG02", "RG03", "RG04", "RG05"],
    "Description": [
        "ID Client vide",
        "Age hors referentiel",
        "Genre inconnu",
        "Montant nul",
        "Doublons"
    ],
    "Type": ["Completude", "Format", "Format", "Coherence", "Unicite"],
    "Anomalies": [0, 61761, 515, 52, 0],
    "Criticite": ["Conforme", "Critique", "Modere", "Modere", "Conforme"]
})

bilan_regles.to_csv("../outputs/bilan_regles.csv", index=False)
print("Export 1 OK")
print(bilan_regles)
```

```python
# Export 2 - Fraude par categorie
fraude_categorie = df.groupby("category")["fraud"].agg(["count", "sum", "mean"]).round(4)
fraude_categorie.columns = ["total_transactions", "total_fraudes", "taux_fraude"]
fraude_categorie["taux_fraude"] = (fraude_categorie["taux_fraude"] * 100).round(2)
fraude_categorie = fraude_categorie.reset_index()

fraude_categorie.to_csv("../outputs/fraude_par_categorie.csv", index=False)
print("Export 2 OK")
print(fraude_categorie)
```

```python
# Export 3 - Detail anomalies age
detail_age = pd.DataFrame({
    "Type_anomalie": ["Encodage incorrect", "Clients mineurs", "Age inconnu"],
    "Nombre": [2452, 58131, 1178],
    "Criticite": ["Critique", "Critique", "Modere"]
})

detail_age.to_csv("../outputs/detail_anomalies_age.csv", index=False)
print("Export 3 OK")
print(detail_age)
```
